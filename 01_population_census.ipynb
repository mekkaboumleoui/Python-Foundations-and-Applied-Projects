# TD1 — Analyse démographique à partir de scénarios de recensement

## Présentation
Ce notebook met en pratique les fonctions Python, les conditions, la gestion d'un cas particulier et les tests simples à partir de trois scénarios démographiques.

**Objectif :** calculer le taux de dépendance total et le taux de dépendance des jeunes, puis classer les résultats selon des seuils définis dans l'exercice.

> Les scénarios numériques ci-dessous sont les données de l'exercice. Pour une publication publique, ajouter une référence officielle si les chiffres du contexte RGPH sont présentés comme des statistiques réelles.

## 1. Importation

```python
from typing import Union

Number = Union[int, float]
```

## 2. Définition des fonctions

Le taux de dépendance total est calculé par rapport à la population en âge de travailler. Le cas où cette population est nulle est traité explicitement.

```python
def calculate_total_ratio(youth: Number, working_age: Number, elderly: Number) -> float:
    """Calcule le taux de dépendance total en pourcentage."""
    if working_age == 0:
        raise ValueError("La population en âge de travailler ne peut pas être nulle.")
    if min(youth, working_age, elderly) < 0:
        raise ValueError("Les effectifs de population doivent être positifs ou nuls.")
    return round(((youth + elderly) / working_age) * 100, 2)


def calculate_youth_ratio(youth: Number, working_age: Number) -> float:
    """Calcule le taux de dépendance des jeunes en pourcentage."""
    if working_age == 0:
        raise ValueError("La population en âge de travailler ne peut pas être nulle.")
    if min(youth, working_age) < 0:
        raise ValueError("Les effectifs de population doivent être positifs ou nuls.")
    return round((youth / working_age) * 100, 2)
```

## 3. Données des trois scénarios

Les données sont regroupées dans un dictionnaire pour éviter de répéter les mêmes blocs de calcul.

```python
scenarios = {
    "A": {"youth": 10_928_637, "working_age": 25_944_272, "elderly": 2_395_339},
    "B": {"youth": 1_200_000, "working_age": 1_800_000, "elderly": 250_000},
    "C": {"youth": 400_000, "working_age": 2_500_000, "elderly": 2_500_000},
}
```

## 4. Tests des fonctions

Les tests vérifient les valeurs attendues pour les trois scénarios ainsi que le traitement du cas `working_age = 0`.

```python
expected = {
    "A": (51.36, 42.12),
    "B": (80.56, 66.67),
    "C": (116.00, 16.00),
}

for name, values in scenarios.items():
    total = calculate_total_ratio(**values)
    youth = calculate_youth_ratio(values["youth"], values["working_age"])
    assert total == expected[name][0], f"Erreur scénario {name}: total={total}"
    assert youth == expected[name][1], f"Erreur scénario {name}: jeunes={youth}"

try:
    calculate_total_ratio(12, 0, 6)
except ValueError:
    pass
else:
    raise AssertionError("Le cas working_age = 0 aurait dû déclencher une ValueError.")

print("Tous les tests ont réussi.")
```

## 5. Analyse et classification

Les seuils de classification ci-dessous reprennent la logique de l'exercice. Ils ne constituent pas une norme démographique universelle.

```python
def classify_total_ratio(value: float) -> str:
    if value > 75:
        return "Dépendance totale élevée"
    if value > 50:
        return "Dépendance totale modérée"
    return "Dépendance totale faible"


def classify_youth_ratio(value: float) -> str:
    if value > 45:
        return "Dépendance des jeunes élevée"
    if value <= 30:
        return "Dépendance des jeunes faible"
    return "Dépendance des jeunes modérée"


for name, values in scenarios.items():
    total = calculate_total_ratio(**values)
    youth = calculate_youth_ratio(values["youth"], values["working_age"])
    print(f"Scénario {name}")
    print(f"  Taux de dépendance total : {total:.2f}% — {classify_total_ratio(total)}")
    print(f"  Taux de dépendance des jeunes : {youth:.2f}% — {classify_youth_ratio(youth)}")
    print()
```

# TD2 — Smart Grid Management avec Python

## Présentation
Ce notebook représente un petit système de gestion de compteurs électriques. Il met en pratique les `dataclasses`, les classes, les fonctions et les opérations sur les listes.

Les données sont des exemples pédagogiques et ne représentent pas des relevés réels du réseau électrique.

## 1. Définition de la structure `SmartMeter`

```python
from dataclasses import dataclass

@dataclass
class SmartMeter:
    serial_id: str
    district: str
    kwh_usage: float
    is_solar: bool

    def __post_init__(self):
        if not self.serial_id.strip():
            raise ValueError("L'identifiant du compteur est obligatoire.")
        if self.kwh_usage < 0:
            raise ValueError("La consommation ne peut pas être négative.")
```

## 2. Création et déploiement des compteurs

```python
pilot_unit = SmartMeter("RBT-01", "Technopolis", 120.5, True)
print(f"Compteur {pilot_unit.serial_id} situé à {pilot_unit.district}")


def deploy_meter(district_name: str, serial_id: str = "NEW-UNIT") -> SmartMeter:
    """Crée un nouveau compteur avec une consommation initiale nulle."""
    return SmartMeter(
        serial_id=serial_id,
        district=district_name.strip(),
        kwh_usage=0.0,
        is_solar=False
    )


@dataclass
class OldCounter:
    id: str
    val: float


def upgrade_infrastructure(old_counter: OldCounter, district: str = "Unknown") -> SmartMeter:
    """Convertit un ancien compteur vers le nouveau modèle SmartMeter."""
    return SmartMeter(
        serial_id=old_counter.id,
        district=district,
        kwh_usage=float(old_counter.val),
        is_solar=False
    )
```

## 3. Création et gestion du réseau

```python
rabat_grid = [
    SmartMeter("RBT-02", "Agdal", 87.5, False),
    SmartMeter("RBT-03", "Hay Riad", 115.0, True),
    SmartMeter("RBT-04", "Temara", 45.2, False),
]


def check_status(grid_list: list[SmartMeter]) -> str:
    return "Online" if grid_list else "Offline"


# Ajouter un compteur
new_meter = SmartMeter("RBT-05", "Bab Rouah", 50.0, False)
rabat_grid.append(new_meter)

# Retirer le premier compteur (exemple de retrait pour maintenance)
meter_for_repair = rabat_grid.pop(0)

print("État du réseau :", check_status(rabat_grid))
print("Compteur retiré :", meter_for_repair.serial_id)
print("Dernière consommation enregistrée :", rabat_grid[-1].kwh_usage, "kWh")
```

## 4. Tests simples

```python
new_meter = deploy_meter("Agdal", "RBT-06")
assert new_meter.district == "Agdal"
assert new_meter.kwh_usage == 0.0
assert check_status([]) == "Offline"
assert check_status(rabat_grid) == "Online"

old = OldCounter("OLD-01", 75.0)
upgraded = upgrade_infrastructure(old, "Centre-ville")
assert upgraded.serial_id == "OLD-01"
assert upgraded.kwh_usage == 75.0

try:
    SmartMeter("RBT-ERR", "Test", -1, False)
except ValueError:
    pass
else:
    raise AssertionError("Une consommation négative aurait dû être refusée.")

print("Tous les tests ont réussi.")
```
